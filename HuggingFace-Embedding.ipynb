#### HuggingFace Embedding

##### Hugging Face provides open-source access to thousands of different community-built embedding models (like BGE or E5) that you can run locally for free.

```python
import os
from dotenv import load_dotenv
load_dotenv()
```

```python
os.environ['HUGGINGFACEHUB_API_TOKEN'] = os.getenv("HUGGINGFACEHUB_API_TOKEN")
```

```python
from langchain_huggingface import HuggingFaceEmbeddings
embedding=HuggingFaceEmbeddings(model="all-MiniLM-L6-v2")
```

```python
text="LangChain is a framework for developing applications powered by language models. It can be used for chatbots, Generative Question-Answering (GQA), summarization, and much more."
query_result=embedding.embed_query(text)
```
