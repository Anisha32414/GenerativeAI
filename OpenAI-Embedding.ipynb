#### Here we will use OpenAI embedding method to convert text chunks to vector and save them in vector database Chroma.


```python
import os
from dotenv import load_dotenv
load_dotenv()
```

```python
os.environ["OPENAI_API_KEY"] = os.getenv("OPENAI_API_KEY")
```

```python
from langchain_openai import OpenAIEmbeddings
embeddings=OpenAIEmbeddings(model="text-embedding-3-large")
embeddings
```

```python
text="This is a sample text to generate embeddings using OpenAI's text-embedding-3-large model."
embedding_vector=embeddings.embed_query(text)

embedding_vector
```

```python
len(embedding_vector)
```

```python
embedding_1024=OpenAIEmbeddings(model="text-embedding-3-large",dimensions=1024)
```

```python
embedding_1024
```

```python
text_1024="This is a sample text to generate embeddings using OpenAI's text-embedding-3-large model with 1024 dimensions."
embedding_vector_1024=embedding_1024.embed_query(text_1024)
embedding_vector_1024
```

```python
len(embedding_vector_1024)
```

```python
# Text Loader
from langchain_community.document_loaders import TextLoader
loader=TextLoader("speech.txt")
document=loader.load()
document
```

```python
from langchain_text_splitters import RecursiveCharacterTextSplitter
text_splitter=RecursiveCharacterTextSplitter(chunk_size=50,chunk_overlap=10)
doc=text_splitter.split_documents(document)
doc
```

```python
# vector embedding & store in vector database

from langchain_community.vectorstores import Chroma
db=Chroma.from_documents(doc,embedding_1024,persist_directory="db")
```

```python
db
```

```python
## Querying the vector database and retrieving relevant documents using similarity search
```

```python
query="Today, we live in a world where Artificial Intelligence"
retrieved_result=db.similarity_search(query)
print(retrieved_result)
```
