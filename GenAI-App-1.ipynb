#### Creating Basic GenAI app using LangChain

```python
import os
from dotenv import load_dotenv
load_dotenv()

os.environ["GOOGLE_API_KEY"]=os.getenv("GOOGLE_API_KEY")
os.environ["LANGCHAIN_API_KEY"]=os.getenv("LANGCHAIN_API_KEY")
os.environ["LANGCHAIN_TRACING_V2"]="true"
os.environ["LANGCHAIN_PROJECT"]=os.getenv("LANGCHAIN_PROJECT")
```

```python
## Data Ingension --- from the website we need to scrape the data so we will use beautifulsoup

from langchain_community.document_loaders import WebBaseLoader
```

```python
loader=WebBaseLoader("https://www.ibm.com/think/topics/large-language-models")
docs=loader.load()
print(docs)
```

#### Steps we will follow:-
* Load Data --> Docs
* Docs --> Chunks
* Chunks --> Embeddings
* Embeddings --> vector Embeddings
* vector DB store

```python
from langchain_text_splitters import RecursiveCharacterTextSplitter
splitter=RecursiveCharacterTextSplitter(chunk_size=1000,chunk_overlap=100)
doc=splitter.split_documents(docs)
doc
```

```python
from langchain_community.embeddings import OllamaEmbeddings

embeddings=(OllamaEmbeddings(model="nomic-embed-text"))
embeddings
```

```python
from langchain_community.vectorstores import FAISS
db=FAISS.from_documents(doc,embeddings)
```

```python
db
```

```python
query="This text is broken down into smaller, machine-readable units called tokens, during a process of tokenization."
result=db.similarity_search(query)
print(result[0].page_content)
```

```python
from langchain_google_genai import ChatGoogleGenerativeAI
llm = ChatGoogleGenerativeAI(model="gemini-3.6-flash")
```

```python
from langchain_classic.chains.combine_documents import create_stuff_documents_chain
```

```python
from langchain_core.prompts import ChatPromptTemplate
prompt=ChatPromptTemplate.from_messages([
    ("system","""Answer the following question based on the provided context:

<context>
{context}
</context>
"""
    ),
    ("human", "{input}")
])
chain_doc=create_stuff_documents_chain(llm,prompt)
```

```python
chain_doc
```

```python
from langchain_core.documents import Document
chain_doc.invoke({"input":"LLMs are initially trained with self-supervised learning, a machine learning"
                  ,"context":[Document(page_content="LLMs are initially trained with self-supervised learning, a machine learning technique that uses unlabeled data for supervised learning. ")]})
```
