#### FAISS (Facebook AI Similarity Search) 
It is a highly optimized, open-source library developed by Meta AI Research that specializes in the efficient similarity search and clustering of dense vectors.

```python
from langchain_community.document_loaders import TextLoader
from langchain_community.vectorstores import FAISS
from langchain_community.embeddings import OllamaEmbeddings
from langchain_text_splitters import CharacterTextSplitter
```

```python
loader=TextLoader("speech.txt")
document=loader.load()
text_splitter=CharacterTextSplitter(separator=" ",chunk_size=200,chunk_overlap=10)
docs=text_splitter.split_documents(document)
len(docs)
```

```python
docs
```

```python
embeddings=OllamaEmbeddings(model="nomic-embed-text")
db=FAISS.from_documents(docs,embeddings)
```

```python
db
```

```python
## Quering the database
text="They also make it much easier to experiment with different LLMs"
result=db.similarity_search(text)
result
```

```python
result[0].page_content
```

```python
### Retrieving the result using as_retriever, invoke method

retriever=db.as_retriever()
result=retriever.invoke(text)
```

```python
result
```

#### Similarity search with score


A similarity search with a score matches query data to stored items using vector embeddings. It returns both the matching items and a quantitative score showing how close or far apart the items are in a data space.

```python
doc=db.similarity_search_with_score(text)
doc
```

```python
#### Saving and Loading the database
```

```python
db.save_local("faiss_index")
```

```python
new_db=FAISS.load_local("faiss_index",embeddings,allow_dangerous_deserialization=True)
```

```python
query="LangChain serves as a generic interface for nearly any LLM"
res=new_db.similarity_search_with_score(query)
```
