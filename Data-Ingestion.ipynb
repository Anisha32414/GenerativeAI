### Data Ingestion Technique

#### Text Loader

```python
# Text Loader
from langchain_community.document_loaders import TextLoader
loader=TextLoader("speech.txt")
loader
```

```python
document=loader.load()
document
```

#### PDF Loader

```python
# PDF Loader
from langchain_community.document_loaders import PyPDFLoader
pdf_loader=PyPDFLoader("LangChain.pdf")
```

```python
file=pdf_loader.load()
file
```

#### WebBased Loader

```python
# WebBased Loader

from langchain_community.document_loaders import WebBaseLoader
web_loader=WebBaseLoader("https://cloud.google.com/ai/llms")
result=web_loader.load()
result
```

```python
# Arxiv Loader

from langchain_community.document_loaders import ArxivLoader
docs=ArxivLoader(query="1605.08386",load_max_docs=1).load()
len(docs)
```

```python
# Wikipedia Loader

from langchain_community.document_loaders import WikipediaLoader
docs=WikipediaLoader(query="LangChain",load_max_docs=2).load()
len(docs)
```
