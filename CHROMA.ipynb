#### CHROMA DB
It is an open-source vector database built for AI applications. It stores and queries vector embeddings—numerical representations of text or images—to power semantic search, long-term LLM memory, and RAG (Retrieval-Augmented Generation) pipelines.

```python
## Building a sample vector DB
from langchain_chroma import Chroma
from langchain_community.embeddings import OllamaEmbeddings
from langchain_community.document_loaders import TextLoader
from langchain_text_splitters import RecursiveCharacterTextSplitter
```

```python
loader=TextLoader("speech.txt")
data=loader.load()
text_splitter=RecursiveCharacterTextSplitter(chunk_size=300,chunk_overlap=20)
docs=text_splitter.split_documents(data)
```

```python
docs
```

```python
embedding=OllamaEmbeddings(model="nomic-embed-text")
db=Chroma.from_documents(docs,embedding)
```

```python
db
```

```python
text="Launched by Harrison Chase in October 2022, LangChain enjoyed a meteoric rise to prominence"
query=db.similarity_search(text)
```

```python
query
```

```python
query[0].page_content
```

```python
# Save the vector database to disk
vector_db=Chroma.from_documents(docs,embedding,persist_directory="chroma_db")
```

```python
# Load the vector database from disk
db2=Chroma(persist_directory="chroma_db",embedding_function=embedding)
docs=db2.similarity_search(text)
```

```python
docs
```

```python
## Similarity search score
query="LangChain can facilitate most use cases for LLMs and natural language processing"
docs=db2.similarity_search_with_score(query)
```

```python
docs
```

```python
### Retriever options
retriever=db2.as_retriever()
retriever.invoke(query)[0].page_content
```
