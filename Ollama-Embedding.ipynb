#### Ollama embedding

##### Ollama lets you run open-source embedding models on your own computer for free. OpenAI provides proprietary embedding models through a paid, cloud-based API

```python
from langchain_community.embeddings import OllamaEmbeddings
```

```python
embeddings=(
    OllamaEmbeddings(model="nomic-embed-text")
)
```

```python
embeddings
```

```python
r1=embeddings.embed_documents(
    [
        "Machine Learning is the sub domian of artificial intelligence",
        "Deep Leaning is the sub domian of machine learning",
    ]
)
```

```python
r1
```

```python
len(r1[0])
```

```python
embeddings.embed_query("What is the sub domain of machine learning?")
```
